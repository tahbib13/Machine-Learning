# Ensemble Learning

Ensemble learning is a method where we use many small models instead of just one. Each of these models may not be very strong on its own, but when we put their results together, we get a better and more accurate answer. It's like asking a group of people for advice instead of just one person—each one might be a little wrong, but together, they usually give a better answer.

![alt text](<Screenshot (109).png>)

1. Voting

Voting is used for classification and also regression problems. It involves making predictions using multiple models and
combining them via majority voting (hard voting) or averaging probabilities (soft voting). The final
prediction is the one most models agree on.

2. Bagging (Bootstrap Aggregating)

Bagging reduces variance by training multiple models on different bootstrapped samples of the data.
Each model is trained independently, and their outputs are combined by averaging (for regression)
or majority voting (for classification). Random Forest is a popular bagging method.

3. Boosting

Boosting reduces bias and variance by training models sequentially. Each new model focuses on
the errors made by the previous ones. This improves performance gradually. Popular algorithms
include AdaBoost, Gradient Boosting, and XGBoost.

4. Stacking

Stacking (or stacked generalization) combines multiple base learners and feeds their outputs to a
meta-learner. The meta-learner then makes the final prediction. This method leverages the
strengths of all models for higher performance.

Bias-Variance Tradeoff

In machine learning, bias refers to errors due to overly simplistic assumptions, while variance refers
to errors due to excessive sensitivity to training data fluctuations. Ensemble methods, particularly
bagging and boosting, help balance this tradeoff:
- Bagging reduces variance without increasing bias.
- Boosting reduces both bias and variance.

# Benefits of Ensemble Learning

- Higher accuracy than individual models
- Reduces overfitting (e.g., bagging)
- Corrects weak learners (e.g., boosting)
- Robust to noise and outliers
- Handles bias-variance tradeoff effectively
